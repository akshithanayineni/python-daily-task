1. Create a hybrid inheritance structure with classes A → B and C → D (where D inherits
from both B and C).
Concept
Hybrid Inheritance (Combination of Single + Multiple Inheritance)
What to do
• Create parent class A.
• Create class B inheriting from A.
• Create another class C.
• Create class D inheriting from both B and C.
How to solve
1. Create class A.
2. Add one method.
3. Create class B(A).
4. Create class C.
5. Create class D(B, C).
6. Create an object of D.
7. Call methods from all parent classes.

```python
class animal:
    def sound(Self):
        print("animal makes sound")
class dog(animal):
    def sound(Self):
        print("dog makes sound")
class cat(animal):
    def sound(Self):
        print("cat makes sound")
class snake(dog,cat):
    def sound(self):
        print(" snake makes sound")
s=snake()
d=dog()
c=cat()
s.sound()
d.sound()
c.sound()
```

2. Create Person, Employee, Student, and TeachingAssistant using hybrid inheritance.
Concept
Real-world Hybrid Inheritance
What to do
• Person is the base class.
• Employee inherits Person.
• Student inherits Person.
• TeachingAssistant inherits Employee and Student.
How to solve


Hierarchy
Person
/ \
Employee Student
\ /
TeachingAssistant
Steps
1. Store name in Person.
2. Employee stores salary.
3. Student stores marks.
4. TeachingAssistant stores department.
5. Display all information.

```python
class person:
    def __init__(self, name, age, **kwargs):
        super().__init__(**kwargs)
        self.name = name
        self.age = age
class employee(person):
    def __init__(self, name, age, salary, **kwargs):
        super().__init__(name=name, age=age, **kwargs)
        self.salary = salary
class student(person):
    def __init__(self, name, age, marks, **kwargs):
        super().__init__(name=name, age=age, **kwargs)
        self.marks = marks
class TeachingAssistant(employee, student):
    def __init__(self, name, age, salary, marks, dept):
        super().__init__(
            name=name,
            age=age,
            salary=salary,
            marks=marks
        )
        self.dept = dept

    def display(self):
        print("Name :", self.name)
        print("Age :", self.age)
        print("Salary :", self.salary)
        print("Marks :", self.marks)
        print("Dept :", self.dept)
t = TeachingAssistant("aksh", 21, 40000, 99, "Teaching")
t.display()
```

19. Create a hybrid inheritance structure and apply Duck Typing.
Concept
Hybrid + Duck Typing
What to do
Create hybrid inheritance.
Create another unrelated class with same method.
How to solve
1. Create hybrid hierarchy.
2. Final child has method show().
3. Another unrelated class also has show().
4. One function calls show() for both.

```python
class Person:
    def show(self):
        print("Person details")
class Student(Person):
    pass
class Teacher(Person):
    pass
class Monitor(Student, Teacher):
    def show(self):
        print("Monitor details")
class Guest:
    def show(self):
        print("Guest details")
def display(obj):
    obj.show()
monitor = Monitor()
guest = Guest()
display(monitor)
display(guest)
```
