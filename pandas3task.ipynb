```python
Challenge 20 – Final Interview Project
Your manager gives you a dirty customer dataset.
Prepare the final dataset.
Requirements
Load dataset
Check information
Find missing values
Remove duplicates
Rename
CustomerID
AnnualIncome
CreditScore
Target

Convert
Age

AnnualIncome
Balance

Fill missing values
Remove
Segment
CustomerID
Create new columns
TotalMoney = Balance + AnnualIncome
AgeAfter5Years = Age + 5
MonthlyIncome = AnnualIncome / 12
CreditScoreDouble = CreditScore * 2Display
• Head
• Tail
• Sample
• Describe
• Target counts
• Final information
```

```python
df.info()
print(df.isnull().sum())
df.drop_duplicates(inplace=True)
df.rename(columns={
    "CustomerID": "Customer_ID",
    "AnnualIncome": "Income",
    "CreditScore": "Credit_Score",
    "Target": "Customer_Status"
}, inplace=True)
df["Age"] = df["Age"].fillna(df["Age"].median()).astype(int)
df["Income"] = df["Income"].astype(float)
df["Balance"] = df["Balance"].astype(float)
df["Income"] = df["Income"].fillna(df["Income"].median())
df["Balance"] = df["Balance"].fillna(df["Balance"].mean())
df.drop(columns=["Segment", "Customer_ID"], inplace=True)
df["TotalMoney"] = df["Balance"] + df["Income"]
df["AgeAfter5Years"] = df["Age"] + 5
df["MonthlyIncome"] = df["Income"] / 12
df["CreditScoreDouble"] = df["Credit_Score"] * 2
print("First 5 Rows:")
print(df.head())
print("Last 5 Rows:")
print(df.tail())
print("Random Sample:")
print(df.sample(10))
print("Statistical Summary:")
print(df.describe())
print("Customer Status Counts:")
print(df["Customer_Status"].value_counts())
print("Final Dataset Information:")
df.info()
```
