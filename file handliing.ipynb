15. Read a file and display the longest word in it.

```python
file = open("details.txt", "r")
data = file.read()
words = data.split()
file.close()
longest_word = " "
for word in words:
    if len(word) > len(longest_word):
        longest_word = word
print(longest_word)
```

24. Write a program to encrypt the content of a file (simple character shifting) and save it in
another file.

```python
file = open("input.txt", "r")
data = file.read()
file.close()
encrypted = ""
for i in data:
    encrypted = encrypted + chr(ord(i) + 3)
file = open("encrypted.txt", "w")
file.write(encrypted)
file.close()
```

25. Read a file containing student marks, calculate average, and write the result in another file.

```python
file = open("marks.txt", "r")
marks = file.readlines()
file.close()
if len(marks) > 0:
    total = 0
    for i in marks:
        total = total + int(i)
    average = total / len(marks)
    file = open("result.txt", "w")
    file.write(str(average))
    file.close()
    print("Average saved successfully")
else:
    print("No marks found in file")
```

28. Read a file and write only the unique lines into another file (remove duplicates).

```python
file = open("data.txt", "r")
lines = file.readlines()
file.close()
unique_lines = set(lines)
file = open("unique.txt", "w")
for line in unique_lines:
    file.write(line)
    print(line, end="")
file.close()
```

29. Create a menu-driven program:
• Write to file
• Read file
• Append to file
• Exit

```python
while True:
    print("1. Write")
    print("2. Read")
    print("3. Append")
    print("4. Exit")
    choice = input()
    if choice == "1":
        file = open("data.txt", "w")
        file.write(input())
        file.close()
        print("Written")
    elif choice == "2":
        file = open("data.txt", "r")
        print(file.read())
        file.close()
    elif choice == "3":
        file = open("data.txt", "a")
        file.write(input())
        file.close()
    elif choice == "4":
        break
    else:
        print("invalid choice")
```
