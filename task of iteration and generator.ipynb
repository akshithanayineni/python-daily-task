```python
Problem 8
Create a custom iterator that prints numbers in reverse order from
10 to 1.
```

```python
class reversenumber:
    def __init__(self):
        self.num = 10
    def __iter__(self):
        return self
    def __next__(self):
        if self.num >=1:
            value=self.num
            self.num -= 1
            return value
        else:
            raise StopIteration
obj=reversenumber()
for i in obj:
    print(i)
```

```python
Problem 9
Create a custom iterator that prints the characters of the word
"GENERATOR" one by one.
```

```python
class character:
    def __init__(self):
        self.word = "GENERATOR"
        self.i = 0
    def __iter__(self):
        return self
    def __next__(self):
        if self.i < len(self.word):
            word = self.word[self.i]
            self.i += 1
            return word
        else:
            raise StopIteration
obj=character()
for i in obj:
    print(i)
```

```python
class Numbers:
    def __init__(self):
        self.num = 1
    def __iter__(self):
        return self
    def __next__(self):
        if self.num <= 15:
            self.num += 1
            return self.num - 1
        raise StopIteration
obj = Numbers()
for i in obj:
    print(i)
```

```python
Problem 24
Create a custom iterator that prints all vowels from the string below.
text = "Artificial Intelligence"
```

```python
text = "Artificial Intelligence"
class Vowels:
    def __init__(self):
        self.text = text
        self.index = 0
    def __iter__(self):
        return self
    def __next__(self):
        while self.index < len(self.text):
            ch = self.text[self.index]
            self.index += 1
            if ch.lower() in "aeiou":
                return ch
        raise StopIteration
obj = Vowels()
for i in obj:
    print(i)
```

```python
Problem 25
Create a custom iterator that prints the Fibonacci sequence up to 10
terms.
```

```python
class Fib:
    def __init__(self):
        self.a = 0
        self.b = 1
        self.i = 1
    def __iter__(self):
        return self
    def __next__(self):
        if self.i <= 10:
            value = self.a
            self.a = self.b
            self.b = value + self.b
            self.i += 1
            return value
        else:
            raise StopIteration
obj = Fib()
for x in obj:
    print(x)
```

```python
def fibonacci():
    a = 0
    b = 1
    for i in range(15):
        yield a
        c = a + b
        a = b
        b = c
obj = fibonacci()
for i in obj:
    print(i)
```

```python
Problem 27
Create a generator that yields all prime numbers between 1 and 100.
```

```python
def prime():
    for num in range(2, 101):
        count = 0
        for i in range(1, num + 1):
            if num % i == 0:
                count += 1
        if count == 2:
            yield num
obj = prime()
for i in obj:
    print(i)
```

```python
Problem 28
Create a generator that yields all factors of a given number.
Example Input:
num = 24
```

```python
num = 24
def factors():
    for i in range(1, num + 1):
        if num % i == 0:
            yield i
obj = factors()
for i in obj:
    print(i)
```

```python
Problem 30
Create an infinite generator that yields multiples of 5.
Print only the first 15 values using next().
```

```python
def multiples():
    num = 5
    while True:
        yield num
        num += 5
obj = multiples()
for i in range(15):
    print(next(obj))
```
