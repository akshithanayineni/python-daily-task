11. Create class BankAccount with private __balance. Add deposit and withdraw
methods. Raise custom exception if balance becomes negative.

```python
class NegativeBalanceError(Exception):
    pass
class BankAccount:
    def __init__(self):
        self.__balance = 0
    def deposit(self, amount):
        self.__balance = self.__balance + amount
        print("Balance:", self.__balance)
    def withdraw(self, amount):
        if self.__balance - amount < 0:
            raise NegativeBalanceError("Insufficient Balance")
        self.__balance = self.__balance - amount
        print("Balance:", self.__balance)
b = BankAccount()
b.deposit(1000)
try:
    b.withdraw(1200)
except NegativeBalanceError as e:
    print(e)
```

14. Create class Rectangle with private length and width. Add method to calculate
area and perimeter.

```python
class Rectangle:
    def __init__(self, length, width):
        self.__length = length
        self.__width = width
    def area(self):
        return self.__length * self.__width
    def perimeter(self):
        return 2 * (self.__length + self.__width)
r = Rectangle(10, 5)
print(r.area())
print(r.perimeter())
```

15. Create single level inheritance: Person → Student. Use super() in constructor.

```python
class Person:
    def __init__(self, name):
        self.name = name
class Student(Person):
    def __init__(self, name, roll_no):
        super().__init__(name)
        self.roll_no = roll_no
s = Student("akshi", 101)
print("Name:", s.name)
print("Roll No:", s.roll_no)
```

18. Create abstract class Shape with abstract method area(). Implement it in Circle
and Square using super().

```python
from abc import ABC, abstractmethod
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
class Circle(Shape):
    def __init__(self, radius):
        super().__init__()
        self.radius = radius
    def area(self):
        return 3.14 * self.radius * self.radius
class Square(Shape):
    def __init__(self, side):
        super().__init__()
        self.side = side
    def area(self):
        return self.side * self.side
c = Circle(5)
s = Square(4)
print(c.area())
print(s.area())
```

```python
21. Create class Point and overload + and > operators.
```

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)
    def __gt__(self, other):
        return (self.x + self.y) > (other.x + other.y)
p1 = Point(2, 3)
p2 = Point(4, 5)
p3 = p1 + p2
print(p3.x, p3.y)
print("p1 > p2:", p1 > p2)
```

26. Create a class Student with private marks list.
• Setter should validate each mark (0-100)
• Overload > operator to compare two students based on average
• Raise custom exception if marks list is empty

```python
class EmptyMarksError(Exception):
    pass
class Student:
    def __init__(self):
        self.__marks = []
    def set_marks(self, marks):
        if len(marks) == 0:
            raise EmptyMarksError("Marks list is empty")
        for mark in marks:
            if mark < 0 or mark > 100:
                print("Invalid mark:", mark)
                return
        self.__marks = marks
    def average(self):
        total = 0
        for mark in self.__marks:
            total = total + mark
        return total / len(self.__marks)
    def __gt__(self, other):
        return self.average() > other.average()
s1 = Student()
s2 = Student()
try:
    s1.set_marks([80, 90, 70])
    s2.set_marks([60, 75, 85])
    print("Student 1 Average:", s1.average())
    print("Student 2 Average:", s2.average())
    if s1 > s2:
        print("Student 1 has higher average")
    else:
        print("Student 2 has higher average")
except EmptyMarksError as e:
    print(e)
```
