```python
def cal(a,b,choice):
    if choice == 1:
        return a + b
    elif choice == 2:
        return a - b
    elif choice == 3:
        return a * b
    elif choice == 4:
        return a / b
    else:
        raise Exception("invalid choice")
try:
    a = int(input())
    b = int(input())
    choice = int(input())
    print(cal(a,b,choice))
except ValueError:
    print("Invalid Number")
except ZeroDivisionError:
    print("Cannot divide by zero")
except Exception as e:
    print(e)
```

28. Create a banking system with deposit and withdraw. Raise and handle custom exception
for insufficient balance.

```python
class InsufficientBalanceError(Exception):
    pass
balance = 6000
try:
    choice = input()
    amount = int(input("Enter amount: "))
    if choice == "A":
        balance += amount
        print("Deposit Successful")
        print("Balance:", balance)
    elif choice == "B":
        if amount > balance:
            raise InsufficientBalanceError("Insufficient Balance")
        balance -= amount
        print("Withdrawal Successful")
        print("Balance:", balance)
    else:
        print("Invalid Choice")
except InsufficientBalanceError as e:
    print(e)
```
