23. Create abstract class Shape with abstract method area(). Implement it in Circle and

```python
from abc import ABC, abstractmethod
class shape(ABC):
    @abstractmethod
    def area(self):
        pass
class circle(shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return 3.14 * self.radius * self.radius
class rectangle(shape):
    def __init__(self, length, width):
        super().__init__()
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width
c = circle(3)
r = rectangle(1,2)
print(c.area())
print(r.area())
```

5. Create abstract class Shape with abstract methods area() and perimeter().
Implement Circle and Rectangle using super(). Add a class attribute to count total
shapes created.

```python
from abc import ABC, abstractmethod
class Shape(ABC):
    total_shapes = 0
    def __init__(self):
        Shape.total_shapes += 1
    @abstractmethod
    def area(self):
        pass
    @abstractmethod
    def perimeter(self):
        pass
class Circle(Shape):
    def __init__(self, radius):
        super().__init__()
        self.radius = radius
    def area(self):
        return 3.14 * self.radius * self.radius
    def perimeter(self):
        return 2 * 3.14 * self.radius
class Rectangle(Shape):
    def __init__(self, length, width):
        super().__init__()
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width
    def perimeter(self):
        return 2 * (self.length + self.width)
c = Circle(5)
r = Rectangle(10, 5)
print("Circle Area:", c.area())
print("Circle Perimeter:", c.perimeter())
print("Rectangle Area:", r.area())
print("Rectangle Perimeter:", r.perimeter())
print("Total Shapes:", Shape.total_shapes)
```

9. Create a class ComplexNumber and overload +, -, *, and == operators completely.

```python
class ComplexNumber:
    def __init__(self, real, imag):
        self.real = real
        self.imag = imag
    def __add__(self, other):
        return ComplexNumber(self.real + other.real,
                              self.imag + other.imag)
    def __sub__(self, other):
        return ComplexNumber(self.real - other.real,
                             self.imag - other.imag)
    def __mul__(self, other):
        real = self.real * other.real - self.imag * other.imag
        imag = self.real * other.imag + self.imag * other.real
        return ComplexNumber(real, imag)
    def __eq__(self, other):
        return self.real == other.real and self.imag == other.imag
    def display(self):
        print(self.real, "+", self.imag, "i")
c1 = ComplexNumber(5, 3)
c2 = ComplexNumber(2, 1)
c3 = c1 + c2
c4 = c1 - c2
c5 = c1 * c2
print("Addition:")
c3.display()
print("Subtraction:")
c4.display()
print("Multiplication:")
c5.display()
print("Equal:", c1 == c2)
```

10. Create a class Student with private marks list.
• Add setter with validation (each mark 0-100)
• Calculate average
• Overload > operator to compare two students based on average

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.__marks = []
    def set_marks(self, marks):
        for mark in marks:
            if mark < 0 or mark > 100:
                print("Marks should be between 0 and 100")
                return
        self.__marks = marks
    def get_marks(self):
        return self.__marks
    def average(self):
        total = 0

        for mark in self.__marks:
            total = total + mark

        return total / len(self.__marks)
    def __gt__(self, other):
        return self.average() > other.average()
s1 = Student("Akshitha")
s2 = Student("kams")
s1.set_marks([80, 90, 70])
s2.set_marks([70, 75, 80])
print("Student 1 Average:", s1.average())
print("Student 2 Average:", s2.average())
if s1 > s2:
    print("Student 1 has higher average")
else:
    print("Student 2 has higher average")
```
